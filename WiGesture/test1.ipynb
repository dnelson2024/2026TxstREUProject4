So we will be using the WiGesture dataset. This dataset was chosen because they collected their data using the ESP32 device with one antenna per device. 
- frequncy band: 2.4 Ghz
- bandwith: 20 Mhz (52 subcarriers)
- protocol: 802.11n
- waveform: OFDM
- sampling rate: 100 Hz

The "Dynamic" folder contains various dynamic gestures, such as applause, while the "Static" folder includes static digital gestures ranging from number 1 to 9.  For the dynamic section, each file contains 1 minute of data, while the static section includes files of 10 seconds each.

There were 8 people sampled.

For more information:
https://github.com/RS2002/CSI-BERT/tree/main/WiGesture
